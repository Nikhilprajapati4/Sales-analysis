### 2. How are monthly sales changing?


```python
import matplotlib.pyplot as plt

quary = """SELECT
    TO_CHAR(
        DATE_TRUNC('month', o.order_purchase_timestamp::TIMESTAMP),
        'YYYY-MM'
    ) AS month,
    ROUND(SUM(op.payment_value)::NUMERIC, 2) AS revenue
FROM olist_orders_dataset o
JOIN olist_order_payments_dataset op
    ON o.order_id = op.order_id
WHERE o.order_status = 'delivered'
GROUP BY DATE_TRUNC('month', o.order_purchase_timestamp::TIMESTAMP)
ORDER BY DATE_TRUNC('month', o.order_purchase_timestamp::TIMESTAMP); """
revenue = pd.read_sql(quary, engine)
revenue.head()

# plt.bar(revenue['month'] , revenue['revenue'] , )
# plt.xlabel('months')
# plt.xticks(rotation=45)   # try 45 or 90 degrees
# plt.tight_layout()        # adjusts spacing so labels fit
# plt.show()
```

### 3.Which product categories generate the most revenue?

### 4.Which states and cities contribute the highest sales?


## Highest sales States

```python
## here we calculate the actual revenue as per the goods only here not calculate the any other charges or payment settlement. 

quary1 = """ Select c.customer_state,sum(op.price) as actual_rev 
from olist_customers_dataset c 
              join olist_orders_dataset od 
              on c.customer_id = od.customer_id join olist_order_items_dataset 
              op on od.order_id = op.order_id where od.order_status = 'delivered' 
              group by c.customer_state order by sum(op.price) DESC ;  """ 



## here we calculate revenue as per how money we get from the customer , where shipping charges and other charges or settlements are included
quary = """ SELECT
    c.customer_state,
    round(SUM(op.payment_value::NUMERIC),2) AS revenue
FROM olist_customers_dataset c
JOIN olist_orders_dataset o
    ON c.customer_id=o.customer_id
JOIN olist_order_payments_dataset op
    ON o.order_id=op.order_id
WHERE o.order_status='delivered'
GROUP BY c.customer_state
ORDER BY revenue DESC; """


state_wise1 = pd.read_sql(quary1,engine)

state_wise = pd.read_sql(quary,engine)
new  = pd.concat([state_wise,state_wise1['actual_rev']] , axis = 1)
new.head()
```

```python
## here we calculate the actual revenue as per the goods only here not calculate the any other charges or payment settlement. 

quary1 = """ Select c.customer_city,sum(op.price) as actual_rev 
from olist_customers_dataset c 
              join olist_orders_dataset od 
              on c.customer_id = od.customer_id join olist_order_items_dataset 
              op on od.order_id = op.order_id where od.order_status = 'delivered' 
              group by c.customer_city order by sum(op.price) DESC ;  """ 



## here we calculate revenue as per how money we get from the customer , where shipping charges and other charges or settlements are included
quary = """ SELECT
    c.customer_city,
    round(SUM(op.payment_value::NUMERIC),2) AS revenue
FROM olist_customers_dataset c
JOIN olist_orders_dataset o
    ON c.customer_id=o.customer_id
JOIN olist_order_payments_dataset op
    ON o.order_id=op.order_id
WHERE o.order_status='delivered'
GROUP BY c.customer_city
ORDER BY revenue DESC; """


city_wise1 = pd.read_sql(quary1,engine)

city_wise = pd.read_sql(quary,engine)
new  = pd.concat([city_wise,city_wise1['actual_rev']] , axis = 1)
new.head()
```

### 5.What is the Average Order Value (AOV)?

```python
quary = """SELECT
    ROUND(AVG(order_total),2) AS average_order_value
FROM
(
    SELECT
        o.order_id,
        SUM(op.payment_value :: numeric) AS order_total
    FROM olist_orders_dataset o
    JOIN olist_order_payments_dataset op
        ON o.order_id=op.order_id
    WHERE o.order_status='delivered'
    GROUP BY o.order_id
) t; """
average_ov = pd.read_sql(quary, engine)
average_ov.head()

```

### 7. Which Months Have Peak Sales?

```python
quary = """SELECT
    TO_CHAR(TO_TIMESTAMP(order_purchase_timestamp,'YYYY-MM-DD HH24:MI:SS'),'Month') AS month_name,
    ROUND(SUM(payment_value :: numeric),2) AS revenue
FROM olist_orders_dataset o
JOIN olist_order_payments_dataset p
ON o.order_id=p.order_id
WHERE order_status='delivered'
GROUP BY month_name
ORDER BY revenue DESC ; """
average_ov = pd.read_sql(quary, engine)
average_ov.head()

```

### 8. Which Products Sell the Most?

```python
quary = """ select ct.product_category_name_english , count(*) as total_sale from olist_order_items_dataset oi 
            inner join olist_orders_dataset od on oi.order_id = od.order_id 
            join olist_products_dataset op on op.product_id = oi.product_id
            join product_category_name_translation ct on ct.product_category_name = op.product_category_name
            where od.order_status = 'delivered'
            group by ct.product_category_name_english 
            order by total_sale DESC"""



total_sales = pd.read_sql(quary ,engine)
total_sales.head()
```

```python
quary = """
SELECT
    o.order_id,
    o.order_purchase_timestamp,
    o.order_status,

    c.customer_unique_id,
    c.customer_city,
    c.customer_state,

    oi.product_id,
    p.product_category_name,

    oi.seller_id,
    s.seller_city,
    s.seller_state,

    oi.price,
    oi.freight_value,

    pay.payment_type,
    pay.payment_installments,
    pay.payment_value,

    r.review_score,

    o.order_delivered_customer_date,
    o.order_estimated_delivery_date

FROM olist_orders_dataset o
JOIN olist_customers_dataset c
    ON o.customer_id = c.customer_id
JOIN olist_order_items_dataset oi
    ON o.order_id = oi.order_id
JOIN olist_products_dataset p
    ON oi.product_id = p.product_id
JOIN olist_sellers_dataset s
    ON oi.seller_id = s.seller_id
LEFT JOIN olist_order_payments_dataset pay
    ON o.order_id = pay.order_id
LEFT JOIN olist_order_reviews_dataset r
    ON o.order_id = r.order_id;"""


summury = pd.read_sql(quary,engine)
summury.to_sql("summury",con=engine,if_exists="replace",index=False)
```
